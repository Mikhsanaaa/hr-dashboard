```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../raw/WA_Fn-UseC_-HR-Employee-Attrition.csv')

# Lihat 5 baris pertama
df.head()
```

Cek Informasi Dataset

```python
# Melihat tipe data dan memastikan tidak ada missing values
df.info()
```

Cek Statistik Dasar

```python
# Melihat ringkasan statistik untuk kolom angka
df.describe()
```

Menjawab Pertanyaan Bisnis #1 (Distribusi Attrition)

```python
# Cek jumlah karyawan yang resign vs bertahan
print(df['Attrition'].value_counts())
print("\nPersentase:")
print(df['Attrition'].value_counts(normalize=True) * 100)

# Visualisasi
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Attrition', hue='Attrition', palette='Set2', legend=False)
plt.title('Distribusi Karyawan (Attrition)')
plt.show()
```

Menjawab Pertanyaan Bisnis #1 (Turnover per Departemen)

```python
# Visualisasi attrition berdasarkan departemen
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Department', hue='Attrition', palette='Set2')
plt.title('Attrition per Departemen')
plt.xticks(rotation=15)
plt.show()
```

Menjawab Pertanyaan Bisnis #2 (Pengaruh Lembur/OverTime)

```python
# Visualisasi attrition berdasarkan OverTime
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='OverTime', hue='Attrition', palette='Set2')
plt.title('Pengaruh OverTime terhadap Attrition')
plt.show()
```
